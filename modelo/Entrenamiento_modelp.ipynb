## MODELO PREDICTORIO

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from joblib import dump
```

```python
# Seleccionar las características (X) y la variable objetivo (y)
X = df_casas.drop(columns=['price'])  # 'precio' es la variable que queremos predecir
y = df_casas['price']
```

```python
# Dividir los datos en conjuntos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Crear el modelo
model = RandomForestRegressor(random_state=42)

# Entrenar el modelo
model.fit(X_train, y_train)

```

```python
# Predecir en el conjunto de prueba
y_pred = model.predict(X_test)

# Evaluar el modelo
mae = mean_absolute_error(y_test, y_pred)
print(f"Error absoluto medio (MAE): {mae}")
```

```python
# Ejemplo: predecir el precio de una nueva propiedad
nueva_propiedad = X_test.iloc[7:8]  # Usar datos del conjunto de prueba como ejemplo
prediccion = model.predict(nueva_propiedad)
print(f"El precio predicho para la nueva propiedad es: {prediccion[0]}")
```

```python
nueva_propiedad
```

```python
# Calcular R^2
r2 = model.score(X_test, y_test)
print(f"Coeficiente de determinación (R^2): {r2}")
```

```python
dump(model, 'modelo_entrenado_analisis_casas.joblib')
print("Modelo guardado exitosamente.")
```
